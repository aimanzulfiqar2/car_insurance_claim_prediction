# Identify features types
  ##The dataset contains:
- Numerical features: measurable numerical values
- Categorical features: categories or codes representing groups
- Binary features: features with two possible states
- Structured string features: text containing useful information that may require extraction
- Identifier: uniquely identifies each record
- Target: the variable we want to predic

```python
# Numerical features
numerical_features = [
    "policy_tenure",
    "age_of_car",
    "age_of_policyholder",
    "population_density",
    "airbags",
    "displacement",
    "cylinder",
    "turning_radius",
    "length",
    "width",
    "height",
    "gross_weight"
]

# Categorical features
categorical_features = [
    "area_cluster",
    "make",
    "segment",
    "model",
    "fuel_type",
    "engine_type",
    "rear_brakes_type",
    "transmission_type",
    "steering_type",
    "gear_box",
    "ncap_rating"
]

# Binary features
binary_features = [
    "is_esc",
    "is_adjustable_steering",
    "is_tpms",
    "is_parking_sensors",
    "is_parking_camera",
    "is_front_fog_lights",
    "is_rear_window_wiper",
    "is_rear_window_washer",
    "is_rear_window_defogger",
    "is_brake_assist",
    "is_power_door_locks",
    "is_central_locking",
    "is_power_steering",
    "is_driver_seat_height_adjustable",
    "is_day_night_rear_view_mirror",
    "is_ecw",
    "is_speed_alert"
]

# Structured string features
structured_string_features = [
    "max_torque",
    "max_power"
]

# Identifier
identifier = ["policy_id"]

# Target
target = ["is_claim"]
```

### Observations

Feature types were identified based on the meaning of the variables, not only their Pandas data types.

- `make` has an `int64` data type, but its values (1–5) represent different vehicle manufacturers/makes. Therefore, it is treated as a categorical feature rather than a numerical feature.
- `gear_box` has an `int64` data type, with values 5 and 6. These represent gearbox categories, so it is treated as a categorical feature.
- `policy_id` is an identifier and does not represent a meaningful predictive feature.
- `is_claim` is the target variable, representing whether an insurance claim was made.
- `max_torque` and `max_power` are structured string features because they contain numerical information within text.
